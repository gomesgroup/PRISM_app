```python
import numpy as np
import matplotlib.pyplot as plt
# Experimental params
T_lock = 5.0            # min
t_scan = 0.0667         # min per scan
SNR_min = 10
kappa = 11363           # sensitivity constant
f = 0.75                # fraction remaining
# Concentrations from 0.05 to 20 mM
C_mM = np.logspace(-1.3, 1.3, 200)  # 10^(-1.3)=0.05, 10^(1.3)=20
C = C_mM / 1000                     # convert to M
# Calculate scans, delay, and k_max
n_req = (SNR_min/(kappa * C))**2
T_delay = T_lock + n_req * t_scan
k_max = ((1/f) - 1) / (C * T_delay)
# Plot
plt.figure(figsize=(8,6))
plt.loglog(C_mM, k_max, color='teal', lw=2)
plt.xlabel('Initial Concentration C₀ (mM)')
plt.ylabel('Max Detectable k (M⁻¹·min⁻¹)')
plt.title('Fastest Observable Second-Order Rate vs. Concentration')
plt.grid(True, which='both', ls='--')
plt.tight_layout()
plt.show()
```
